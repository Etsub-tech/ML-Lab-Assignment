```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    confusion_matrix, ConfusionMatrixDisplay, classification_report,
    RocCurveDisplay
)

df = pd.read_csv("classification_escalation.csv")
print("Shape:", df.shape)
display(df.head())
print("\nInfo:")
print(df.info())
print("\nMissing values:")
print(df.isnull().sum())
print("\nTarget prevalence:")
print(df["escalated"].value_counts(normalize=True).rename("proportion"))
```

```python
X = df.drop(columns=["escalated"])
y = df["escalated"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

num_cols = ["complexity_score", "team_experience_years", "customer_tenure_months"]
cat_cols = ["priority", "channel"]

num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(drop="first", handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])

print("Train rows:", len(X_train), "| Test rows:", len(X_test))
```

```python
models = {
    "Majority Baseline": DummyClassifier(strategy="most_frequent", random_state=42),
    "Logistic Regression": LogisticRegression(random_state=42, max_iter=1000),
    "LDA": LinearDiscriminantAnalysis(),
    "k-NN (k=15)": KNeighborsClassifier(n_neighbors=15)
}

fitted = {}
rows = []

for name, clf in models.items():
    pipe = Pipeline([("prep", preprocessor), ("model", clf)])
    pipe.fit(X_train, y_train)
    fitted[name] = pipe
    
    y_pred = pipe.predict(X_test)
    roc_auc = roc_auc_score(y_test, pipe.predict_proba(X_test)[:, 1]) if hasattr(pipe, "predict_proba") else np.nan
    
    rows.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc
    })

comp_df = pd.DataFrame(rows)
display(comp_df.round(4))
```

```python
lr_pipe = fitted["Logistic Regression"]
y_lr_pred = lr_pipe.predict(X_test)
y_lr_prob = lr_pipe.predict_proba(X_test)[:, 1]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test, y_lr_pred, display_labels=["Not Escalated (0)", "Escalated (1)"],
    ax=axes[0], cmap="Blues", colorbar=False
)
axes[0].set_title("Logistic Regression Confusion Matrix (t = 0.50)")

RocCurveDisplay.from_predictions(y_test, y_lr_prob, ax=axes[1], name="Logistic Regression")
axes[1].plot([0, 1], [0, 1], "k--", label="Random Guess")
axes[1].set_title("Logistic Regression ROC Curve")
axes[1].legend(loc="lower right")

plt.tight_layout()
plt.show()
```

```python
thresholds = [0.20, 0.30, 0.40, 0.50, 0.60, 0.70]
t_rows = []

for t in thresholds:
    y_t = (y_lr_prob >= t).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, y_t).ravel()
    t_rows.append({
        "Threshold": t,
        "Precision": precision_score(y_test, y_t, zero_division=0),
        "Recall": recall_score(y_test, y_t, zero_division=0),
        "F1": f1_score(y_test, y_t, zero_division=0),
        "FP": fp,
        "FN": fn,
        "Cost (5*FN + FP)": 5 * fn + fp
    })

thresh_df = pd.DataFrame(t_rows)
display(thresh_df.round(4))
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
knn_pipe = Pipeline([("prep", preprocessor), ("model", KNeighborsClassifier())])

grid = GridSearchCV(
    knn_pipe,
    param_grid={"model__n_neighbors": [5, 9, 15, 21, 31]},
    scoring="f1",
    cv=cv
)
grid.fit(X_train, y_train)

cv_df = pd.DataFrame(grid.cv_results_)[
    ["param_model__n_neighbors", "mean_test_score", "std_test_score", "rank_test_score"]
].rename(columns={"param_model__n_neighbors": "k", "mean_test_score": "Mean CV F1", "std_test_score": "Std CV F1", "rank_test_score": "Rank"})

display(cv_df.round(4))
print("Best k:", grid.best_params_["model__n_neighbors"], "| Best CV F1:", round(grid.best_score_, 4))
```

# Answers to the Questions

# 1. Why is accuracy alone potentially misleading?
Accuracy just tells you the share of predictions that came out right ((TP + TN) / Total), which can look great even when a model misses the cases you actually care about. Here, roughly 79.33% of tickets are not escalated (0) and only 20.67% are escalated (1). A simple baseline that predicts 0 for every single ticket still hits 79.56% accuracy on the test set, yet catches 0% of escalated tickets (Recall = 0.0000, F1 = 0.0000), making it useless in practice.

## 2. Which classifier has the strongest recall? Which has the strongest precision/F1?
At the default 0.50 threshold on the test set:
- Strongest Recall: Logistic Regression and LDA tie at 0.1957 (catching 9 of the 46 escalated tickets), beating k-NN (k=15) at 0.1522 and the majority baseline at 0.0000.
- Strongest Precision and F1: Logistic Regression comes out ahead on both, with 0.8182 precision (versus LDA at 0.7500 and k-NN at 0.7000) and 0.3158 F1 (versus LDA at 0.3103 and k-NN at 0.2500).

# 3. What changes when the threshold is lowered?
Lowering the threshold (for example, from 0.50 down to 0.20) means the model flags a ticket as escalated (1) at a lower predicted probability:
- False negatives (FN) go down (37 at 0.50 down to 10 at 0.20), so recall jumps from 0.1957 to 0.7826.
- False positives (FP) go up (2 at 0.50 up to 49 at 0.20), which pulls precision down (0.8182 to 0.4235), while F1 rises at 0.20 (0.5496) and 0.30 (0.5773) compared to 0.50 (0.3158).

# 4. Which threshold minimizes 5×FN + FP on the test set, and why might this estimate be optimistic if used repeatedly for tuning?
Out of the thresholds tested (0.20, 0.30, 0.40, 0.50, 0.60, 0.70), 0.20 gives the lowest test cost at 99 (5×10 + 49 = 99). If you keep adjusting or picking a threshold based on how it scores on the test set, you end up overfitting to test noise. A cutoff that looks great on one 225-row test set might just be riding random quirks in that split and perform worse on new tickets.

## 5. Explain prior, likelihood, posterior, and evidence using the ticket scenario.
- Prior P(escalated=1): How likely a ticket is to be escalated before you look at its details (about 20.7% overall in this dataset).
- Likelihood P(x | escalated=1): How likely you are to see a ticket's feature pattern x (like a high complexity_score, high priority, or less experienced team) given that the ticket actually ended up escalated.
- Evidence P(x): How common that feature pattern x is across all tickets (escalated or not), acting as the normalizing factor in Bayes' rule.
- Posterior P(escalated=1 | x): The updated probability that this specific ticket needs escalation once you factor in the prior, likelihood, and evidence.

## 6. With posterior P(escalated | x) = 0.30, false-positive cost 1, and false-negative cost 5, compare expected cost of escalating vs. not escalating, and derive the cost-sensitive threshold.
With p = P(escalated=1 | x) = 0.30, C_FP = 1, and C_FN = 5:
- Expected cost of Escalating (y_hat = 1):
  $$E[\text{Cost} \mid \text{Escalate}] = p \cdot 0 + (1 - p) \cdot C_{FP} = 1 - p = 1 - 0.30 = 0.70$$
- Expected cost of Not Escalating (y_hat = 0):
  $$E[\text{Cost} \mid \text{Do Not Escalate}] = p \cdot C_{FN} + (1 - p) \cdot 0 = 5p = 5(0.30) = 1.50$$
- Since 0.70 is lower than 1.50, escalating is the cheaper choice.
- Cost-sensitive threshold (p*): Escalate whenever the expected cost of escalating is less than or equal to not escalating:
  $$(1 - p) C_{FP} \le p C_{FN} \implies p \ge \frac{C_{FP}}{C_{FP} + C_{FN}}$$
  Plugging in C_FP = 1 and C_FN = 5:
  $$p^* = \frac{1}{1 + 5} = \frac{1}{6} \approx 0.1667$$
  So escalating whenever P(escalated | x) >= 0.1667 minimizes expected cost.

## 7. Which model would you recommend for deployment, and what further checks are required?
I would go with Logistic Regression using a lower cost-aware threshold around 0.20 (or near 0.17, verified via cross-validation). Before rolling it out, we should check probability calibration, validate the threshold using cross-validation instead of test-set peeking, test on real ticket history, track data drift and missingness, check error rates across channels, priorities, and customer tenure groups, and keep a human override in place alongside SLA monitoring.


# Recommendation (300 to 500 Words)

For handling ticket escalation triage, I recommend deploying Logistic Regression paired with a lower threshold around 0.20 (close to the theoretical cost-optimal threshold of ~0.17, which should be confirmed via cross-validation). On the test set, Logistic Regression gives the strongest overall ranking (ROC-AUC = 0.811) and the highest F1-score (0.316) at the default 0.50 cutoff, outperforming LDA (ROC-AUC = 0.809, F1 = 0.310), k-NN (ROC-AUC = 0.748, F1 = 0.250), and the majority baseline (accuracy = 0.796, F1 = 0.000).

With the default 0.50 threshold, Logistic Regression plays it way too safe. It only catches 19.6% of tickets that actually needed escalation (recall = 0.196, leaving 37 false negatives versus just 2 false positives). Under the 5×FN + 1×FP penalty, that adds up to a steep error cost of 187. In real support operations, letting a high-risk ticket slip through unescalated is much worse than escalating an extra ticket. Missed escalations can mean broken SLAs, angry customers, or churn, whereas a false positive mostly just costs a bit of extra review time. Dropping the threshold to 0.20 lifts recall to 78.3% (bringing false negatives down from 37 to 10, while false positives rise from 2 to 49) and nearly cuts the test-set error cost in half to 99, while lifting F1 to 0.550.

The data is synthetic and fairly small (900 rows, 20.7% escalated), and several columns have missing values (team_experience_years has 31, customer_tenure_months has 22, and priority has 18).
